```python
import pandas as pd
import numpy as np

df = pd.read_csv("dataset/retail_store_inventory.csv")

df["Date"] = pd.to_datetime(df["Date"])

df = df.sort_values(
    ["Store ID", "Product ID", "Date"]
).reset_index(drop=True)

print(df.shape)
df.head()
```

```python
features = [
    "Store ID",
    "Product ID",
    "Category",
    "Region",
    "Inventory Level",
    "Units Sold",
    "Units Ordered",
    "Price",
    "Discount",
    "Weather Condition",
    "Holiday/Promotion",
    "Seasonality"
]

print(features)
```

```python
excluded_features = [
    "Demand Forecast"
]

print("Excluded features:", excluded_features)
```

```python
X = df[features]

print("Feature shape:", X.shape)
print("Features:")
print(X.columns.tolist())
```
